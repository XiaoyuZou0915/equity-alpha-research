```python
from pathlib import Path
import pandas as pd

# VS Code notebook is normally run from the project root
data_path = Path("../data/raw/ohlcv_sample_2020_2024.csv")

prices = pd.read_csv(
    data_path,
    header=[0, 1],
    index_col=0,
    parse_dates=True
)

print("Shape:", prices.shape)
print("Columns:")
print(prices.columns)

display(prices.head())
```

```python
print("Date range:")
print("Start:", prices.index.min())
print("End:", prices.index.max())

print("\nIndex checks:")
print("DatetimeIndex:", isinstance(prices.index, pd.DatetimeIndex))
print("Sorted ascending:", prices.index.is_monotonic_increasing)
print("Duplicate dates:", prices.index.duplicated().sum())

print("\nMissing values by field and ticker:")
missing_counts = prices.isna().sum()
print(missing_counts.to_string())

print("\nNon-positive Close prices:")
print((prices["Close"] <= 0).sum())

print("\nNegative Volume observations:")
print((prices["Volume"] < 0).sum())
```

```python
ticker = "XOM"

comparison = pd.DataFrame({
    "Close": prices["Close"][ticker],
    "Adj Close": prices["Adj Close"][ticker],
    "Dividends": prices["Dividends"][ticker],
    "Stock Splits": prices["Stock Splits"][ticker]
})

comparison["Adjustment Ratio"] = (
    comparison["Adj Close"] / comparison["Close"]
)

display(comparison.head())
display(comparison.tail())
```

```python
# Day 3 — Step 1.1: Corporate Actions Audit

# Check available corporate-action fields
print("Available price fields:")
print(prices.columns.get_level_values("Price").unique().tolist())

# Extract dividends and stock splits
dividends = prices["Dividends"]
splits = prices["Stock Splits"]

# Count corporate-action events by ticker
action_summary = pd.DataFrame({
    "Dividend Days": (dividends > 0).sum(),
    "Split Days": (splits != 0).sum(),
    "Total Dividends Per Share": dividends.sum()
})

display(action_summary)

# Show XOM dividend events
xom_dividend_events = dividends.loc[dividends["XOM"] > 0, "XOM"]

print("\nXOM dividend events:")
display(xom_dividend_events.head(10))

# Show all stock split events
split_events = splits[splits.ne(0).any(axis=1)]

print("\nStock split events:")
display(split_events)
```

```python
# Day 3 — Step 2.1: AAPL stock split analysis

ticker = "AAPL"

aapl = pd.DataFrame({
    "Close": prices["Close"][ticker],
    "Adj Close": prices["Adj Close"][ticker],
    "Stock Splits": prices["Stock Splits"][ticker]
})

# Calculate daily returns
aapl["Close Return"] = aapl["Close"].pct_change()
aapl["Adj Return"] = aapl["Adj Close"].pct_change()

# Inspect the period around the split
split_window = aapl.loc["2020-08-26":"2020-09-03"]

display(split_window.round(4))
```

```python
# Day 3 — Step 2.2: XOM dividend analysis

ticker = "XOM"

xom = pd.DataFrame({
    "Close": prices["Close"][ticker],
    "Adj Close": prices["Adj Close"][ticker],
    "Dividends": prices["Dividends"][ticker]
})

# Calculate daily returns
xom["Close Return"] = xom["Close"].pct_change()
xom["Adj Return"] = xom["Adj Close"].pct_change()

# A simple dividend-inclusive one-day return
# Only valid when the dividend and prices use compatible share units
xom["Dividend-Inclusive Return"] = (
    (xom["Close"] + xom["Dividends"])
    / xom["Close"].shift(1)
    - 1
)

# Inspect one dividend event
dividend_window = xom.loc["2020-02-06":"2020-02-12"]

display(dividend_window.round(5))

# Compare return measures on the event date
event_date = "2020-02-10"

display(
    xom.loc[
        event_date,
        [
            "Dividends",
            "Close Return",
            "Adj Return",
            "Dividend-Inclusive Return"
        ]
    ]
)
```

```python
# Day 3 — Step 3.2: PIT adjustment experiment

# Hypothetical prices observed before a future stock split
price_start = 180.0
price_end = 200.0

# Momentum using original historical prices
momentum_original = price_end / price_start - 1

# A future 2-for-1 split leads the vendor to
# multiply both historical prices by 0.5
adjustment_factor = 0.5

price_start_adjusted = price_start * adjustment_factor
price_end_adjusted = price_end * adjustment_factor

momentum_adjusted = (
    price_end_adjusted / price_start_adjusted - 1
)
print(f"Original momentum: {momentum_original:.6%}")
print(f"Adjusted momentum: {momentum_adjusted:.6%}")
print(
    "Difference:",
    round(momentum_adjusted - momentum_original, 12)
)
```

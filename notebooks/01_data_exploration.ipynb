```python
import sys
import pandas as pd
import yfinance as yf

print("Python:", sys.executable)
print("Pandas:", pd.__version__)
print("yfinance:", yf.__version__)
```

```python
tickers = ["AAPL", "MSFT", "JPM", "XOM", "WMT"]
start_date = "2020-01-01"
end_date = "2025-01-01"
```

```python
prices_raw = yf.download(
    tickers=tickers,
    start=start_date,
    end=end_date,
    interval="1d",
    auto_adjust=False,
    actions=True,
    group_by="column",
    progress=False,
    threads=False
)

print("Shape:", prices_raw.shape)
print("Columns:", prices_raw.columns)
print(prices_raw.head())
```

```python
from pathlib import Path
import pandas as pd

# VS Code notebook is normally run from the project root
data_path = Path("../data/raw/ohlcv_sample_2020_2024.csv")

prices = pd.read_csv(
    data_path,
    header=[0, 1],
    index_col=0,
    parse_dates=True
)

print("Shape:", prices.shape)
print("Columns:")
print(prices.columns)

display(prices.head())
```

```python
print("Date range:")
print("Start:", prices.index.min())
print("End:", prices.index.max())

print("\nIndex checks:")
print("DatetimeIndex:", isinstance(prices.index, pd.DatetimeIndex))
print("Sorted ascending:", prices.index.is_monotonic_increasing)
print("Duplicate dates:", prices.index.duplicated().sum())

print("\nMissing values by field and ticker:")
missing_counts = prices.isna().sum()
print(missing_counts.to_string())

print("\nNon-positive Close prices:")
print((prices["Close"] <= 0).sum())

print("\nNegative Volume observations:")
print((prices["Volume"] < 0).sum())
```

```python
ticker = "XOM"

comparison = pd.DataFrame({
    "Close": prices["Close"][ticker],
    "Adj Close": prices["Adj Close"][ticker],
    "Dividends": prices["Dividends"][ticker],
    "Stock Splits": prices["Stock Splits"][ticker]
})

comparison["Adjustment Ratio"] = (
    comparison["Adj Close"] / comparison["Close"]
)

display(comparison.head())
display(comparison.tail())
```

```python
print("Total missing values:", int(prices.isna().sum().sum()))

print(
    "Non-positive Close:",
    int((prices["Close"] <= 0).sum().sum())
)

print(
    "Negative Volume:",
    int((prices["Volume"] < 0).sum().sum())
)

print(
    "Zero Volume:",
    int((prices["Volume"] == 0).sum().sum())
)
```
